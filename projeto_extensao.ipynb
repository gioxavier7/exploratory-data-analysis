## 📥 5. Importação e Leitura dos Dados

#### 📌 Conceito: DataFrame

Um DataFrame é uma estrutura de dados bidimensional (linhas e colunas), semelhante a uma planilha do Excel.

No Python, a principal biblioteca para trabalhar com DataFrames é o `pandas`.

#### Importação das Bibliotecas

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

plt.rcParams["figure.figsize"] = (10,6)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 12
```

##### ✔ O que este código faz?

- **pandas:** manipulação de dados

- **matplotlib:** criação de gráficos

- **seaborn:** visualizações estatísticas mais sofisticadas

- **sns.set():** melhora a estética dos gráficos

#### Leitura do Dataset

```python
# Leitura do arquivo CSV
df = pd.read_csv("jobs_in_data.csv")

# Visualização das primeiras linhas
df.head()
```

#### ✔ O que este código faz?

- `pd.read_csv()` carrega o arquivo CSV

- Cria um DataFrame chamado df

- `df.head()` mostra as primeiras 5 linhas

- Permite verificar se os dados foram carregados corretamente

#### Remoção de Duplicados

```python
# Removendo possíveis registros duplicados
df = df.drop_duplicates()
```

#### ✔ O que este código faz?

Remoção de possíveis registros duplicados para garantir integridade da análise.
Essa etapa é recomendada em qualquer processo de EDA.

#### Visualização das Últimas Linhas

```python
df.tail()
```

#### ✔ O que este código faz?

- Mostra as últimas 5 linhas do dataset

- Permite verificar estrutura e possíveis inconsistências no final do arquivo

## 📊 7. Exploração Inicial dos Dados

#### Estatísticas Descritivas (Variáveis Numéricas)

```python
df.describe()
```

#### ✔ O que este código faz?

Gera estatísticas como:

- Média

- Desvio padrão

- Valor mínimo

- Quartis

- Valor máximo

Aplicável apenas às variáveis numéricas

#### Análise de Correlação

```python
plt.figure()
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Matriz de Correlação")
plt.show()
```

#### ✔ O que este código faz?

A matriz de correlação mede o grau de relação linear entre variáveis numéricas.
Valores próximos de 1 ou -1 indicam forte correlação.
Valores próximos de 0 indicam baixa relação linear.

#### Estatísticas para Variáveis Categóricas

```python
df.describe(include='object')
```

#### ✔ O que este código faz?

Mostra:

- Frequência

- Categoria mais comum

- Número de categorias únicas

#### Estatísticas Avançadas (Skewness e Kurtosis)

```python
df['salary_in_usd'].skew()
df['salary_in_usd'].kurtosis()
```

#### ✔ O que este código faz?

- Skewness mede a assimetria da distribuição.
- Kurtosis indica a presença de caudas pesadas (outliers extremos).
Essas métricas aprofundam a análise estatística da variável salário.

## 📈 8. Análise Visual Inicial

#### Distribuição dos Salários em USD

```python
plt.figure()
plt.hist(df['salary_in_usd'], bins=30)
plt.title("Distribuição dos Salários em USD")
plt.xlabel("Salário (USD)")
plt.ylabel("Frequência")
plt.show()
```

#### ✔ O que este código faz?


- Cria um histograma

- Mostra como os salários estão distribuídos

Permite observar:

- Assimetria

- Concentração de valores

- Possíveis outliers

#### Contagem por Nível de Experiência

```python
plt.figure()
sns.countplot(data=df, x='experience_level', order=('Entry-level', 'Mid-level', 'Senior', 'Executive'))
plt.title("Distribuição por Nível de Experiência")
plt.xlabel("Nível de Experiência")
plt.ylabel("Quantidade")
plt.show()
```

#### ✔ O que este código faz?


- Cria gráfico de contagem

- Mostra frequência de cada nível de experiência

- Permite entender a composição do dataset

#### Top 10 Países com Mais Registros

```python
top_countries = df['company_location'].value_counts().head(10)

plt.figure()
sns.barplot(x=top_countries.values, y=top_countries.index)
plt.title("Top 10 Países com Mais Vagas")
plt.xlabel("Vagas")
plt.ylabel("Países")
plt.show()
```

#### ✔ O que este código faz?


#### Salário Médio por Nível de Experiência

```python
plt.figure()
sns.boxplot(data=df, x='experience_level', y='salary_in_usd')
plt.title("Salário em USD por Nível de Experiência")
plt.show()
```

#### ✔ O que este código faz?



- Cria boxplot

- Mostra: Mediana, Quartis, Dispersão, Outliers

- Permite comparar salários entre níveis de experiência
